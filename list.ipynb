#### List Comprehension
Basic Syntax [expression for item in iterable]

with conditional logic [expression for item in iterable if condition]

Nested list Comprehension [expression for item1 in iterable1 for item2 in iterable2]

```python
#x**2 : what we want as output 
#for x in range 10 : what kind ofoperation we want to perform
[x**2 for x in range (10)]
```

```python
#Basic List Comprehension
sqaure=[num**2 for num in range(10)]
print(sqaure)
```

```python
#Even Numbers
lst=[]
for i in range(10):
    if i%2 ==0:
        lst.append(i)
print(lst)
    
```

```python
even_numbers=[num for num in range(10) if num%2==0] #what we want as o/p; kind of operation we want to perform ; if condition
print(even_numbers)
```

```python
#Nested List Comprehension
lst1 = [1,2,3,4]
lst2=['a','b','c','d']
pair=[[i,j] for i in lst1 for j in lst2]
print(pair)
```

# Example 1: To Do List

```python
##To do List
To_do_list=["Buy Groceries","Buy a pen","Clean the house","Pay bills"]
To_do_list.append("Schedule Meeting")
To_do_list.append("GO for Run")
To_do_list.remove('Clean the house')
if 'Pay bills' in To_do_list:
    print("Don't forget to pay utility bills")

print("To do List Remaining")
for task in To_do_list:   #To print in this manner 
    print(f"-{task}")
print(To_do_list) ##To print inside []
```

# Example 2 : Organizing Student Grades 

```python
#Organizing student grades
grades=[85,92,78,94]

#Adding a new grade 
grades.append(95)

# calculating the average grade
Average_grade = sum(grades)/len(grades)
print(f"Average Grades: {Average_grade: .2f}")

#Finding the highest and lowest grade 
highest_grade = max(grades)
lowest_grade = min(grades)
print(f"Highest Grade : {highest_grade: .2f}")
print(f"Lowest Grade : {lowest_grade: .2f}")
```

# Example 3: Managing an Inventory

```python
#Managing Inventory
inventory=["Apples","Banana","Oranges","Grapes"]

#Adding a new item
inventory.append("Strawberries")

#Removing an item that is out of stock
inventory.remove("Banana")

#Checking if an item is in stock
item = "Oranges"
if item in inventory:
    print(f"{item} are in Stock")
else :
    print(f"{item} are out of Stock")

#Printing Inventory
print("Inventory List: ")
for item in inventory:
    print(f"-{item}")
```

# Example 4 : Collecting User Feedback

```python
# Collecting user feedback
feedback = ["Great service!", "Very satisfied", "Could be better", "Excellent experience"]

# Adding new feedback
feedback.append("Not happy with the service")

# Counting specific feedback
positive_feedback_count = sum(1 for comment in feedback if "great" in comment.lower() or "excellent" in comment.lower())

print(f"Positive Feedback Count: {positive_feedback_count}")

# Printing all feedback
print("User Feedback:")
for comment in feedback:
    print(f"- {comment}")
```
