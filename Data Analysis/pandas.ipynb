```python
## Pandas : Powerful data manipulation library in Python widely used for data manipulation and data cleaning.
## It provides two primary data structures : Series and Data Frame 
## Series: 1D array like object that can hold any datatype. It is similar to a column in object.
## Data Frame : 2D,multiple rows and columns, size mutable, and potentially 
## heterogeneous tabular data structure with labeled axes ( rows and columns)
```

```python
import pandas as pd
```

```python
data = [1,2,3,4,5]
series = pd.Series(data)
print("Series: \n", series)
print(type(series))
```

```python
## Create a series from Dictionary
data = {'a':1,'b':2,'c':3}
pd.Series(data)
```

```python
data = [10,20,30]
index = ['a','b','c']
pd.Series(data,index=index)
```

```python
## Create a dataframe from dictionary of list 
data = {
    'Name':['Rose','Jack','Dawson'],
    'age':[19,20,22],
    'City':['Banglore','New York','Los Vegas']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a Data Frame from a list of Dictionaries 
data = [
    {'Name':'Rose','age':32, 'City':'Banglore'},
    {'Name':'Jack','age':16, 'City':'Los Vegas'},
    {'Name':'Dawson','age':24, 'City':'New York'},
    {'Name':'Daniel','age':26, 'City':'New York'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
