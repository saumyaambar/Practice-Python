####  Tuples : Tuples are ordered collection of items that are immutable. 
#### They are similar to lists but their immutability makes them different.

```python
## creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))
```

```python
lst = list()
print(type(lst))
tpl=tuple()
print(type(tpl))
```

```python
## Add elements
## List to Tuple 
numbers = tuple([1,2,3,4,5,6])
print(numbers)
```

```python
## Tuple to list
numbers = list((1,2,3,4,5,6))
print(numbers)
```

```python
## mixed tuple
mixed_tuple = (1,"Hello world",3.14,True)
print(mixed_tuple)
```

```python
## Accessing Tuple elemnts 
numbers = (1,2,2,3,4,5,6)
numbers[0]
```

```python
print(numbers[2])
print(numbers[-1])
```

```python
numbers[0:4]
```

```python
numbers[::-1]
```

```python
## Tuple Operations

## Concatenation
concatenation_tuple = mixed_tuple + numbers
print(concatenation_tuple)
```

```python
## Multiply
mixed_tuple*3
```

```python
numbers*3
```

```python
## Immutable nature of tuples
## Tuples are immutable, meaning their elements cannot be changed once assigned.
lst = [1,2,3,4,5]
print(lst)
lst[1]="Krish"
print(lst)
```

```python
tpl=(1,2,3,4,5)
print(tpl)
tpl[1]="Krish"
print(tpl)
```

```python
#### So how will mutate? First change the tuple into list then mutate then change the list back to tuple.
```

```python
print(numbers.index(4))
```
