```python
# Dictionaries

### Unordered colletion of items. They store data in key-value pairs. Keys must be unique and immutable (eg. strings,number,tuples) 
### while values can be of any type
```

```python
## Creating Dictionaies
empty_dict = {}
print(type(empty_dict))
```

```python
empty_dict = dict() # another way ; dict() class
empty_dict
```

```python
# Key: Value (key must be unique )
student = {"name":"Krish", "age":20,"marks":98} 
print(student)
print(type(student))
```

```python
## Accessing Dictionary Elements
student = {"name":"Krish", "age":20,"grade":'A'} 
print(student['grade'])
print(student['age'])

## Accessing using get() method
print(student.get('grade'))
print(student.get('grade','B'))
print(student.get('last_name'))
print(student.get('last_name','Not available'))
```

```python
## Modifying Dictionaries Element 
## Dictonaires are mutable, so u can add ,update and delete.
print(student)
```

```python
student = {"name":"Krish", "age":20,"grade":'A'}
student["Address"] = "India"   #added a new key and value
print(student)
student["age"] = 33  #update value for the key
print(student)
```

```python
## to delete
del student['grade']  # delete key and value pair
print(student)
```

```python
## Dictinary methods
keys = student.keys()  #get all the keys.
print(keys)
values=student.values()  #get all the values.
print(values)

items=student.items() #get all key value pairs.
print(items)
```

```python
## shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student['name']= 'krish1'
print(student)
print(student_copy)
```

```python
## Iterating Over Dictionaries 
## u can use loops to iterate over dictionaries,key,values or items.


## Iterating over keys
for key in student.keys():
    print(key)
```

```python
## Iterating over values 
for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested Dictionaries
student= {
    "student1":{'name':'Krish','age':32,'grade':'A'},
    "student2":{'name':'Krish','age':32,'grade':'A'}
}
print(student)
```

```python
## Access Nested Dictionaries
student= {
    "student1":{'name':'Krish','age':32,'grade':'A'},
    "student2":{'name':'Krish','age':32,'grade':'A'}
}
print(student['student1']['name'])
print(student['student1']['age'])
```

```python
student.items()
```

```python
## Iterating over nested Dictionaries
student= {
    "student1":{'name':'Krish','age':32,'grade':'A'},
    "student2":{'name':'Krish','age':32,'grade':'A'}
}
for student_id,student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension
squares= {x:x**2 for x in range(5)}
print(squares)
```

```python
## Conditional Dictionary Comprehension
evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```

## Practical Examples

```python
## Use a dictionary to count the frequency of the elements in list

numbers = [1,2,2,3,3,3,4,4,5,6,7,7,7,8,9,10]
freq = {}

for number in numbers:
    if number in freq:
        freq[number]+=1
    else:
        freq[number]=1
print(freq)
```

```python
## Merge Dictionary 

dict_1={"a":1,"b":2}
dict_2={"c":3,"d":4}
merge_dict={**dict_1,**dict_2}
print(merge_dict)
```
