```python
#Conditional Loops 
# if-else statements
#elif

#Nested conditional loops

num = int(input("Enter a number"))
if num>=0:
    print("Positive Number")
    if num%2==0:
        print("Even Number")
    else:
        print("Odd Number")
else:
    print("Negative Number")
    if num%2==0:
        print("Even Number")
    else:
        print("Odd Number")
```

```python
#Leap Year
# if divisible by 4 and not divisible by 100 then a leap year
# if divisible by 100 then it also must be divisible by 400 to be a leap year 
year = int(input("Enter the year :"))
if year%4==0:
    if year%100!=0:
         print(year , "is a Leap Year")
    else:
        if year%400==0:
            print(year, " is a leap year")
        else: 
            print(year, "is not a leap year")
```
